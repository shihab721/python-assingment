```python
#Tuple
emp = []
for i in range(20):
    name = input("Enter employee name: ")
    emp.append(name)

employees = tuple(emp)

# 1. Find each name and its frequency
print("\nEmployee Name and Frequency:")
for name in set(employees):
    print(name, ":", employees.count(name))

# 2. Remove duplicate names and find the number of distinct names
distinct = tuple(set(employees))
print("\nTuple after removing duplicates:")
print(distinct)
print("Number of distinct names =", len(distinct))

# 3. Print the employee name having maximum frequency
max_freq = 0
for name in set(employees):
    if employees.count(name) > max_freq:
        max_freq = employees.count(name)
        max_name = name
print("\nEmployee with Maximum Frequency:")
print(max_name, "-", max_freq, "times")

# 4. Sort the tuple in alphabetical order
sorted_tuple = tuple(sorted(employees))
print("\nEmployees in Alphabetical Order:")
print(sorted_tuple)

# 5. Search for a specific employee name
search = input("\nEnter employee name to search: ")
if search in employees:
    print(search, "exists in the tuple.")
else:
    print(search, "does not exist in the tuple.")
```
