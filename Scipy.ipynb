```python
import numpy as np
# Define two 3x3 matrices
A = np.array([
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
])
B = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 1]
])
# i. Inverse of matrix A
inverse_A = np.linalg.inv(A)
print("Matrix A:", A)
print("\nMatrix B:", B)
print("\nInverse of Matrix A:", inverse_A)
# ii. Determinant of matrix B
det_B = np.linalg.det(B)
print("\nDeterminant of Matrix B:", det_B)
# iii. Matrix multiplication A.B
result = np.dot(A, B)
print("\nA.B:")
print(result)
```

```python
import numpy as np
from scipy.linalg import solve
A = np.array([
    [2, 3],
    [4, 5]
])
B = np.array([8, 14])
solution = solve(A, B)
print("Value of x =", solution[0])
print("Value of y =", solution[1])
```

```python
import numpy as np
import scipy.linalg

#4x4 matrix
A = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8],
    [2, 4, 6, 8],
    [9, 10, 11, 12]
])
# Transpose of matrix
transpose = A.T

# Rank of matrix
rank = np.linalg.matrix_rank(A)

print("Original Matrix:\n", A)

print("\nTranspose of Matrix:\n",transpose)

print("\nRank of Matrix:", rank)
```

```python
import numpy as np
from scipy.linalg import eig
A = np.array([
    [4, 1, 0, 0],
    [1, 3, 0, 0],
    [0, 0, 2, 1],
    [0, 0, 1, 2]
])
eigenvalues, eigenvectors = eig(A)
print("Matrix A:\n",A)
print("\nEigenvalues:\n", eigenvalues)
print("\nEigenvectors:\n",eigenvectors)
```

```python
import numpy as np
from scipy.linalg import lu
A = np.array([
    [4, 1, 0, 0],
    [1, 3, 0, 0],
    [0, 0, 2, 1],
    [0, 0, 1, 2]
])
P, L, U = lu(A)
print("Matrix A:\n",A)
print("\nP Matrix:\n",P)
print("\nL Matrix:\n",L)
print("\nU Matrix:\n",U)
```

```python
import numpy as np
from scipy.linalg import qr, svd, lstsq
# Square matrix for QR and SVD
A = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 10]
])
# i. QR Decomposition
Q, R = qr(A)
print("QR Decomposition")
print("Q =\n",Q)
print("R =\n",R)
# ii. SVD
U, S, Vt = svd(A)
print("\nSVD")
print("U =\n",U)
print("Singular Values =\n",S)
print("V Transpose =\n",Vt)
# iii. Least Squares (LSC)
A2 = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
b = np.array([5, 11, 17])
x, residuals, rank, singular_values = lstsq(A2, b)
print("\nLeast Squares Solution")
print("Solution =", x)
print("Residuals =", residuals)
print("Rank =", rank)
print("Singular Values =", singular_values)
```
