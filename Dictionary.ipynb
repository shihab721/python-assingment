```python
#Dictionary
employees = {
    "E1": {"empName": "Aman", "designation": "Manager", "department": "HR", "salary": 60000},
    "E2": {"empName": "Riya", "designation": "Developer", "department": "IT", "salary": 50000},
    "E3": {"empName": "Rahul", "designation": "Analyst", "department": "Finance", "salary": 55000},
    "E4": {"empName": "Neha", "designation": "Designer", "department": "Marketing", "salary": 45000},
    "E5": {"empName": "Pooja", "designation": "Engineer", "department": "Production", "salary": 65000}
}
# i. Print the record of employee with empId E1
print("Record of Employee E1:")
print(employees["E1"])

# ii. Print the department of employee E4
print("\nDepartment of Employee E4:")
print(employees["E4"]["department"])

# iii. Print the record of employee having maximum salary
max_salary = 0
max_emp = ""

for empid in employees:
    if employees[empid]["salary"] > max_salary:
        max_salary = employees[empid]["salary"]
        max_emp = empid

print("\nEmployee with Maximum Salary:")
print(max_emp, employees[max_emp])

# iv. Insert a new employee record
employees["E6"] = {
    "empName": "Ankit",
    "designation": "Tester",
    "department": "QA",
    "salary": 48000
}

print("\nEmployee Dictionary after inserting new record:")
for empid in employees:
    print(empid, ":", employees[empid])
```
