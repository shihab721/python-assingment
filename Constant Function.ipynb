```python
#Shapes
import math
class Shape:
    def __init__(self, radius):
        self.radius = radius

class Circle(Shape):
    def cal_area(self):
        return math.pi * self.radius * self.radius

class Sphere(Shape):
    def cal_volume(self):
        return (4 / 3) * math.pi * self.radius ** 3

c = Circle(7)
print("Area of Circle:", c.cal_area())

s = Sphere(7)
print("Volume of Sphere:", s.cal_volume())
```

```python
#Triangles
import math
class Triangle:
    def __init__(self, side1, side2, side3, angle1, angle2, angle3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3
        self.angle1 = angle1
        self.angle2 = angle2
        self.angle3 = angle3


class EquilateralTriangle(Triangle):
    def cal_area(self):
        return (math.sqrt(3) / 4) * self.side1 ** 2

    def find_angle(self):
        print("Angle 1:", math.tan(math.radians(self.angle1)))
        print("Angle 2:", math.tan(math.radians(self.angle2)))
        print("Angle 3:", math.tan(math.radians(self.angle3)))


class ScaleneTriangle(Triangle):
    def cal_perimeter(self):
        return self.side1 + self.side2 + self.side3

    def cal_area(self):
        s = self.cal_perimeter() / 2
        return math.sqrt(
            s * (s - self.side1) *
            (s - self.side2) *
            (s - self.side3)
        )


# Equilateral Triangle
e = EquilateralTriangle(6, 6, 6, 60, 60, 60)

print("Equilateral Triangle")
print("Area:", round(e.cal_area()))
e.find_angle()

print()

# Scalene Triangle
s = ScaleneTriangle(3, 4, 5, 60, 60, 60)

print("Scalene Triangle")
print("Perimeter:", s.cal_perimeter())
print("Area:", round(s.cal_area()))
```
