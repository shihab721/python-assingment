```python
#Lambda Function
student = {
    101: {"Name":"Rohan", "Department":"AI", "Marks":85},
    102: {"Name":"Aman", "Department":"CSE", "Marks":92},
    103: {"Name":"Priya", "Department":"IT", "Marks":78},
    104: {"Name":"Sneha", "Department":"AI", "Marks":95},
    105: {"Name":"John", "Department":"IT", "Marks":88},
}

#Sort accoring to marks (highest to lowest)
sorted_student=dict(sorted(student.items(), key=lambda x: x[1]["Marks"],reverse=True))
print("Students sorted according to thier narks:")
for roll,data in sorted_student.items():
    print(roll,data)

#Student who scored maximum marks
maximum=max(student.items(), key=lambda x: x[1]["Marks"])
print("\nStudent with maximum marks :\n",maximum)

#Average marks of the student
average=sum(map(lambda x:x["Marks"], student.values())) / len(student)
print("\nAverage marks:",average)

#Student scored than the average marks
print("\nStudent scored more than the average:")
for roll, data in student.items():
    if(lambda x:x["Marks"]>average) (data):
        print(roll,data)
```
