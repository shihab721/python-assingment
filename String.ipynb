```python
#String
text="Python Programming"

#Display Python
print(text[:6])

#Display Programming
print(text[7:])

#Check java is present or not, if not include it
if "Java" not in text:
    text=text.replace("Python","Python Java ")
print(text)

#Length of new string
print("Length:",len(text))

#Count number of words in the string
print("No. of words:",len(text.split()))

#Capitalize each word
print(text.upper())

#Remove all spaces
print(text.replace(" ",""))

#Count the frequency of "A,P,R,M"
print("The frequency of (A,P,R,M) are: ")
for ch in "APRM":
    print(ch,"=",text.upper().count(ch))
```
