```python
#Constructor
class Student:
    def __init__(self,name,dept,roll_no):
        self.name=name
        self.dept=dept
        self.roll_no=roll_no

    def show(self):
        print("Name:",self.name)
        print("Deptartment:",self.dept)
        print("Roll No.:",self.roll_no)
        print()

s1=Student("Rohan","CSE",101)
s2=Student("Aman","AI ML",102)
s3=Student("Priya","EE",103)
s4=Student("John","IT",104)
s5=Student("Sneha","CSE",105)

s1.show()
s2.show()
s3.show()
s4.show()
s5.show()
```
